## MFPT

### Downloading

```python
# Importings
from damavand.damavand.datasets.downloaders import ZipDatasetDownloader
from damavand.damavand.datasets.digestors import MFPT

# Downloading the dataset
downloader = ZipDatasetDownloader(addresses['MFPT'])
downloader.download_extract('/content/MFPT.zip', '/content/MFPT/')
```

### Mining

```python
# Mining the dataset
mfpt = MFPT('/content/MFPT/', [
    'baseline_1.mat',
    'InnerRaceFault_vload_1.mat',
    'InnerRaceFault_vload_2.mat',
    'InnerRaceFault_vload_4.mat',
    'InnerRaceFault_vload_7.mat',
    'OuterRaceFault_1.mat',
    'OuterRaceFault_vload_1.mat',
    'OuterRaceFault_vload_2.mat',
    'OuterRaceFault_vload_4.mat',
    'OuterRaceFault_vload_7.mat',
])
mining_params = {
    97656: {'win_len': 19200, 'hop_len': 4800},
    48828: {'win_len': 9600, 'hop_len': 2400},
}
mfpt.mine(mining_params)
```

### Resampling

```python
mfpt_data_48 = pd.concat(mfpt.data[48828]).reset_index(drop = True)
mfpt_data_97 = pd.concat(mfpt.data[97656]).reset_index(drop = True)
mfpt_data_97
```

```python
from damavand.damavand.augmentations import resampling
mfpt_data_97_downsampled = resampling(mfpt_data_97.iloc[:, :-4], 9600)
mfpt_data_97_downsampled = pd.concat([mfpt_data_97_downsampled, mfpt_data_97.iloc[:, -4:]], axis = 1)

import numpy as np
np.all(
    mfpt_data_97_downsampled.iloc[:, -4:].values ==
    mfpt_data_97.iloc[:, -4:].values
)
```

```python
mfpt_data_97_downsampled
```

### Aggregating the whole data

```python
mfpt_data_df = pd.concat([mfpt_data_48, mfpt_data_97_downsampled]).reset_index(drop = True)
mfpt_data_df
```

```python
mfpt_data_df["state"].value_counts()
```

```python
mfpt_data_df.to_csv("/content/drive/MyDrive/phme26/data/mfpt_data.csv", index = True)
```

## KAIST

### Downloading the dataset

```python
downloader = ZipDatasetDownloader(addresses['KAIST'])
downloader.download_extract('/content/KAIST.zip', '/content/KAIST/')
```

### Mining the dataset

```python
from damavand.damavand.datasets.digestors import KAIST

# Mining the dataset (using only two channels out of four)
kaist = KAIST('/content/KAIST/', os.listdir('/content/KAIST/'), [1])
mining_params = {
    'win_len': 4800,
    'hop_len': 4800,
}
kaist.mine(mining_params)
```

```python
kaist_df = pd.concat(df for df in kaist.data[1] if df["state"].value_counts().index[0] in ["BPFI", "Normal", "BPFO"]).reset_index(drop = True)
kaist_df
```

```python
kaist_df["state"].value_counts()
```

```python
kaist_df.to_csv("/content/drive/MyDrive/phme26/data/kaist_data.csv", index = True)
```
